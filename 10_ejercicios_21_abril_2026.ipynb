```python
notas = {
    "Ana": 4.5,
    "Luis": 3.8,
    "Carlos": 4.2
}

def promedio_general(dic):
    return sum(dic.values()) / len(dic)

def promedio_estudiante(dic, nombre):
    return dic[nombre]

print("Promedio general:", promedio_general(notas))
print("Nota de Ana:", promedio_estudiante(notas, "Ana"))
```

```python
palabra = "magia"
conteo = {}

for letra in palabra:
    if letra in conteo:
        conteo[letra] += 1
    else:
        conteo[letra] = 1

print(conteo)
```

```python
lista = ["gema", "oro", "gema", "madera", "oro", "oro", "gema"]
conteo = {}

for elemento in lista:
    if elemento in conteo:
        conteo[elemento] += 1
    else:
        conteo[elemento] = 1

print(conteo)
```
