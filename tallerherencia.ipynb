##Vehiculo y Carro

```python
class Vehiculo:
    def __init__(self, marca):
        self.marca = marca

    def mostrar_marca(self):
        print("Marca:", self.marca)


class Carro(Vehiculo):
    def __init__(self, marca, modelo):
        super().__init__(marca)
        self.modelo = modelo


carro1 = Carro("Toyota", "Corolla")

carro1.mostrar_marca()
print("Modelo:", carro1.modelo)
```

##Persona y Estudiante

```python
class Persona:
    def __init__(self, nombre):
        self.nombre = nombre


class Estudiante(Persona):
    def __init__(self, nombre, curso):
        super().__init__(nombre)
        self.curso = curso


est1 = Estudiante("Valentina", "Programación")

print("Nombre:", est1.nombre)
print("Curso:", est1.curso)
```

##Animal y perro

```python
class Animal:
    def __init__(self, nombre):
        self.nombre = nombre


class Perro(Animal):
    def __init__(self, nombre, raza):
        super().__init__(nombre)
        self.raza = raza


perro1 = Perro("Max", "Labrador")

print("Nombre:", perro1.nombre)
print("Raza:", perro1.raza)
```

##Ropa y Camisa

```python
class Ropa:
    def __init__(self, color):
        self.color = color


class Camisa(Ropa):
    def __init__(self, color, talla):
        super().__init__(color)
        self.talla = talla


camisa1 = Camisa("Azul", "M")

print("Color:", camisa1.color)
print("Talla:", camisa1.talla)
```

##Persona y Profesor

```python
class Persona:
    def __init__(self, nombre, edad):
        self.nombre = nombre
        self.edad = edad

    def saludar(self):
        print("Hola, mucho gusto")


class Profesor(Persona):
    def __init__(self, nombre, edad, materia):
        super().__init__(nombre, edad)
        self.materia = materia


prof1 = Profesor("Carlos", 40, "Matemáticas")

print("Nombre:", prof1.nombre)
print("Edad:", prof1.edad)
print("Materia:", prof1.materia)

prof1.saludar()
```

##Cuenta y Cuenta corriente

```python
class Cuenta:
    def __init__(self, titular, saldo):
        self.titular = titular
        self.saldo = saldo

    def mostrar_saldo(self):
        print("Saldo:", self.saldo)


class CuentaCorriente(Cuenta):
    def __init__(self, titular, saldo, numero_cuenta):
        super().__init__(titular, saldo)
        self.numero_cuenta = numero_cuenta


cuenta1 = CuentaCorriente("Valentina", 500000, "123456789")

print("Titular:", cuenta1.titular)
print("Número de cuenta:", cuenta1.numero_cuenta)

cuenta1.mostrar_saldo()
```

##Empleado y Diseñador

```python
class Empleado:
    def __init__(self, nombre, salario):
        self.nombre = nombre
        self.salario = salario

    def trabajar(self):
        print("El empleado está trabajando")


class Diseñador(Empleado):
    def __init__(self, nombre, salario, programa):
        super().__init__(nombre, salario)
        self.programa = programa


dis1 = Diseñador("Laura", 2500000, "Photoshop")

print("Nombre:", dis1.nombre)
print("Salario:", dis1.salario)
print("Programa que utiliza:", dis1.programa)

dis1.trabajar()
```

##Figura y Triangulo

```python
class Figura:
    def mostrar_figura(self):
        print("Figura: Triángulo")


class Triangulo(Figura):
    def __init__(self, base, altura):
        self.base = base
        self.altura = altura

    def calcular_area(self):
        return (self.base * self.altura) / 2


tri1 = Triangulo(10, 5)

tri1.mostrar_figura()
print("Área del triángulo:", tri1.calcular_area())
```

##Deportista y Nadador

```python
class Deportista:
    def __init__(self, nombre, deporte):
        self.nombre = nombre
        self.deporte = deporte

    def entrenar(self):
        print("El deportista está entrenando")


class Nadador(Deportista):
    def __init__(self, nombre, deporte, estilo):
        super().__init__(nombre, deporte)
        self.estilo = estilo


nad1 = Nadador("Andrés", "Natación", "Mariposa")

print("Nombre:", nad1.nombre)
print("Deporte:", nad1.deporte)
print("Estilo de natación:", nad1.estilo)

nad1.entrenar()
```
