##Clase persona

```python
class Persona:
    def __init__(self, edad):
        self.__edad = edad

    def get_edad(self):
        return self.__edad

    def set_edad(self, edad):
        if edad > 0:
            self.__edad = edad
        else:
            print("La edad debe ser mayor a 0")


p1 = Persona(18)

print("Edad:", p1.get_edad())

p1.set_edad(20)

print("Nueva edad:", p1.get_edad())
```

##Validar saldo negativo

```python
class Cuenta:
    def __init__(self, saldo):
        self.__saldo = saldo

    def depositar(self, dinero):
        if dinero > 0:
            self.__saldo += dinero
            print("Depósito realizado")
        else:
            print("No se permiten depósitos negativos")

    def mostrar_saldo(self):
        print("Saldo:", self.__saldo)


cuenta1 = Cuenta(500)

cuenta1.depositar(200)
cuenta1.depositar(-50)

cuenta1.mostrar_saldo()
```

##Retiro de dinero

```python
class Cuenta:
    def __init__(self, saldo):
        self.__saldo = saldo

    def depositar(self, dinero):
        if dinero > 0:
            self.__saldo += dinero

    def retirar(self, dinero):
        if dinero > self.__saldo:
            print("Fondos insuficientes")
        elif dinero <= 0:
            print("Cantidad inválida")
        else:
            self.__saldo -= dinero
            print("Retiro realizado")

    def mostrar_saldo(self):
        print("Saldo:", self.__saldo)


cuenta1 = Cuenta(1000)

cuenta1.retirar(300)
cuenta1.retirar(800)

cuenta1.mostrar_saldo()
```

##Estudiante

```python
class Estudiante:
    def __init__(self):
        self.__nota = 0

    def asignar_nota(self, nota):
        if 0 <= nota <= 5:
            self.__nota = nota
        else:
            print("La nota debe estar entre 0 y 5")

    def mostrar_nota(self):
        print("Nota:", self.__nota)


est1 = Estudiante()

est1.asignar_nota(4.5)

est1.mostrar_nota()
```

##Temperatura

```python
class Temperatura:
    def __init__(self, celsius):
        self.__celsius = celsius

    def get_temperatura(self):
        return self.__celsius

    def set_temperatura(self, celsius):
        if -100 <= celsius <= 100:
            self.__celsius = celsius
        else:
            print("Temperatura fuera del rango lógico")


temp1 = Temperatura(25)

print("Temperatura:", temp1.get_temperatura())

temp1.set_temperatura(30)

print("Nueva temperatura:", temp1.get_temperatura())
```

##Usuario con contraseña

```python
class Usuario:
    def __init__(self, password):
        self.__password = password

    def cambiar_password(self, nueva_password):
        if len(nueva_password) >= 6:
            self.__password = nueva_password
            print("Contraseña cambiada")
        else:
            print("La contraseña debe tener mínimo 6 caracteres")


user1 = Usuario("123456")

user1.cambiar_password("abc123")
```

##vehiculo

```python
class Vehiculo:
    def __init__(self, velocidad):
        self.__velocidad = velocidad

    def acelerar(self, aumento):
        if aumento >= 0:
            self.__velocidad += aumento
        else:
            print("No se permite velocidad negativa")

    def mostrar_velocidad(self):
        print("Velocidad:", self.__velocidad)


v1 = Vehiculo(60)

v1.acelerar(20)

v1.mostrar_velocidad()
```

##Cuenta Bancaria

```python
class CuentaBancaria:
    def __init__(self, saldo):
        self.__saldo = saldo

    def depositar(self, dinero):
        if dinero > 0:
            self.__saldo += dinero
            print("Depósito exitoso")
        else:
            print("No se permiten valores negativos")

    def retirar(self, dinero):
        if dinero <= 0:
            print("Cantidad inválida")
        elif dinero > self.__saldo:
            print("No hay suficiente saldo")
        else:
            self.__saldo -= dinero
            print("Retiro exitoso")

    def get_saldo(self):
        return self.__saldo


cuenta1 = CuentaBancaria(1000)

cuenta1.depositar(500)
cuenta1.retirar(300)

print("Saldo actual:", cuenta1.get_saldo())
```
